**Numpy and Pytorch can be easily linked**

```python
import numpy as np
import torch
```

```python
a = torch.tensor([1,2,3])
print (type(a)) # tensor
```

```python
# conversion of pytorch tensor to numpy array
b = a.numpy()
print (type(b)) # temsor converted to numpy array
```

```python
c = np.array([4,4,6])
print (type(c)) # numpy array
```

```python
d = torch.from_numpy(c)
print (type(d)) # numpy array converted to a tensor in pytorch
```
