**creating tensors in pytorch**

```python
# empty tensor
a = torch.empty(3,4)

# using zeros
b = torch.zeros(3,3)   # intializes all the tensor values with 0

# tensor with ones
c = torch.ones(2,3)

# to create a tensor with random values in the range of 0 to 1
d = torch.rand(2,3)

# to create my own tensor of my own size we use
x = torch.tensor([[1,2,4],[4,58,7]])  # one tensor can only be in a single [] ,multiple dimentions are gvien by [[]...[]]

# print
print ("my own tensor :\n",x)
print ("empty :\n",a)
print ("zeros :\n",b)
print ("ones :\n",c)
print ("rand :\n",d)

```

random tensors in pytorch cannot be reproduced using the rand funcion for this we seed the value using manual_seed() in torch


```python
# to reproduce the same randonmly generated value we use a seed
torch.manual_seed(291)
x = torch.rand(2,3)
print ("first rand :\n",x )

torch.manual_seed(291)
y = torch.rand(2,3)
print ("second rand :\n",y )
```

```python
#type is a torch tensor object
type (torch.zeros(2,3))

# to find the shape of a tensor in torch we use
print (a.shape)
print (b.shape)
print (c.shape)
print (d.shape)
```

other common ways of creating tensors in pytorch

```python
# arange
# uses a default step size of one and dosent include the endpoint in the tensor
a = torch.arange (1,10) # create the tensor in this range
b = torch.arange (1,10,2) # added a step value of two
c = torch.arange (2.1,4.1, 0.04)

print (a, b, c, sep="\n")
```

```python
# linspace
# this function is used to generate tensors with linearly spaced values and inlcudes the endpoint
a = torch.linspace(20,30,10)
print (a)
```

```python
# eye : to generate an identity matrix of a given shape ( will be used in neural networks )
a =  torch.eye(5)
print(a)
```

```python
# full : initialises the tensor of a given shape with a custom default value
a = torch.full((3,5),-625) # parameters ((<shape>),<value>)
print(a)
```

to create another tensor of the same shape like that of an existing tensor, pytorch has methods like

```python
# to recreate tensors of the same shape
temp = torch.tensor ([[1,2,4],[8,9,6]])

a = torch.empty_like(temp)
b = torch.zeros_like(temp)
c = torch.ones_like(temp)
d = torch.full_like(temp,30)
# e = torch.rand_like(temp)
# this has an issue since rand can only generate values in the range of 0 to 1 in float it cannot generate an integer tensor

print (a,b,c,d, sep = "\n")
```

```python
# tensor.rand_like(<tensor>)
x = torch.tensor([[1,2,3],[8,5,9]])
y = torch.rand (2,3)

# to demonstrate the error
try :
  # rand_like (x) will throw an error because x is an integer array with values not in the range of 0 to 1
  print ("torch.rand_like(x):")
  temp = torch.rand_like(x)
  print (temp)
except Exception as err :
  print ("error:",err)
finally :
  temp = torch.rand_like(y)
  print ("torch.rand_like(y) :",temp, sep="\n")
```
