**Special functions for tensor operations**

```python
import torch

print (torch.__version__)
if torch.cuda.is_available():
  print ("using gpu :", torch.cuda.get_device(0))
else :
  print ("gpu not available using cpu!")
```

```python
torch.manual_seed(100)
f =  torch.rand(4,6)*10
print (f)
```

```python
# log
print (torch.log(f), end="\n\n")
# exp
print (torch.exp(f), end="\n\n")
# sin
print (torch.sin(f), end="\n\n")
# cos
print (torch.cos(f), end="\n\n")
# tan
print (torch.tan(f), end="\n\n")
# sqrt
print (torch.sqrt(f), end="\n\n")
# sigmoid                                                         # IMPORTANT
print (torch.sigmoid(f), end="\n\n")
# relu                                                            # IMPORTANT
print (torch.relu(f), end="\n\n")
# softmax                                                         # IMPORTANT
print (torch.softmax(f, dim=0), end="\n\n")
```

```python
# more on relu
# often used as an activation function in neural networks it has only one purpose
# for any value with a negative value relu changes it to 0

torch.manual_seed(120)
temp = torch.rand(5,5)
torch.inverse(temp)
print (temp, end="\n\n")

c =  ((temp*100)/21) - 2    # random operation to get some megative values
print (c, end="\n\n")

print (c.relu_())
print ("all values with a ngative value will get intialized to 0 with relu")
```
