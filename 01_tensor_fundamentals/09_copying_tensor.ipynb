**copying one tensor value to another in pytorch**

```python
import torch
print(torch.__version__)
print("gpu is available ?:", torch.cuda.is_available())
```

```python
a = torch.rand(3,3)
print(a)
```

```python
# assigning a value with assignment
b = a      # uses call by reference there for any change to a will also change b
```

```python
a[0][0] = 0
print("a=",a)
print("b=",b)
```

```python
# to avoid this we use a function called clone
b = a.clone()
a[2][2] = 0
print("a=",a)
print("b=",b)
```
