**Utilizing available gpu for tensor operations**

```python
import torch
print (torch.__version__)
print ("gpu :",torch.cuda.is_available())
```

```python
# creating tensors on the gpu
# step one
device = torch.device('cuda')
print (device)
```

```python
a = torch.rand((256,256), device=device)    # the parameter ensures that the resultant tensor is created on the gpu
print (a)
```

```python
temp = torch.rand(3,3)
print(temp.device)     # temp is a tensor created on the cpu
print(temp)
```

```python
# moving a tensor to the gpu we simply use the to function with the device attribute
temp.to(device=device)
```
