**Mathematical operations one tensors**

```python
import torch

print (torch.__version__)
if torch.cuda.is_available():
  print ("using gpu :", torch.cuda.get_device(0))
else :
  print ("gpu not available using cpu!")
```

1. Scalar Operations on Tensors

```python
# scalar operations on the tensor x
x = torch.rand(2,4)
print (x, end="\n\n")

# addition
print (x + 7, end="\n\n")

# substraction
print (x - 2, end="\n\n")

# multiplication
print (x * 100, end="\n\n")

# division
print (x / 3, end="\n\n")

# int division
print ((x * 100)//3, end="\n\n")

# mod
print (((x * 100)//3)%2, end="\n\n")

# power
print (x ** 2, end="\n\n")
```

```python
# one example to get random value between 100 use
(x * 100).to(torch.int32).to(torch.float32)
```

2. Element Wise Operaitons

```python
a = torch.rand(3,2)
b = torch.rand_like(a)

# element wise operations on tensors

# addition
print (a + b, end="\n\n")
# substractions
print (a - b , end="\n\n")
# multiplication
print (a * b, end="\n\n")
# division
print (a / b, end="\n\n")
# int division
print ((a * b * 100)//2, end="\n\n")
# mod
print (a%b, end="\n\n")
# power
print (a**b, end="\n\n")
```

```python
# miscellaneous functions
c = torch.tensor ([1,-2,3,-4])
```

```python
# abs
# to  find the absolute values of the tensor, i.e positive equivallent of the numbers
c.abs()
```

```python
# neg
c.neg()
```

```python
d = torch.tensor([1.99, 2.3, 3.7, 4.4])
```

```python
#round
d.round()

# greater than or equal .5 the value becomes ceil and less then .5 the value becomes floor of the current value
```

```python
#ceil
d.ceil()
```

```python
#floor
d.floor()
```

```python
#clamp
torch.clamp(d, min=2, max=3)

# note : the clamp funcitons is used to limit the values of an existing vector in a particualr range
# for all values lower than the the value becomes equal to min here 1.9 becomes 2
# for all values greater than the limit the value becomes equal to the max value here 4.4 becomes 3
```
