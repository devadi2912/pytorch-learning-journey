**Reshaping tensors in pytorch**

```python
import torch
print (torch.__version__)

if torch.cuda.is_available():
  print ("gpu :", torch.cuda.get_device_name(0))
else:
  print("using cpu")
```

reshping tensors in an important operation when dealing with neural networks and training data
some one the common ways to reshape tensors include

```python
a = torch.rand(4,4)
a
```

```python
# reshape
a = a.reshape(2,2,2,2)   # the dimention of the reshaped matri should have the same product as the dimentions of the parent matrix
print (a, end="\n\n")
a = a.reshape(16)
print (a)

# note : the procudt of the dimention of the reshaped tensors should be the same as that of the product of the dimentions of the parent tensor
```

```python
# flatten
b = torch.ones (4,4)
b.flatten()   # brings the tensor to a singel dimentsion
```

```python
a = torch.randint(size=(2,3,4,67), low=0, high=100)
print (a)
```

```python
# permute

# used to reshuffle the dimentions of an array
"""a had dimentions with indices 0,1,2,3"""

print (a.shape)
a = a.permute(2,1,3,0)  # reorder the indices
print (a.shape)
```

```python
# unsqueeze
# used to add a dimentions to the tensor
# often when changing the training data to another set (ex: grey-scale image to rgb image)

a = torch.rand(2,4,3) # single tensor of an rgb image
a = a.unsqueeze(0)      # takes the parameter of the index of the dimention to change
print (a)
```

```python
# squeeze
# does the opposite of squeze removes the dimention form an index
print (a.shape)
print (a.squeeze(0).shape)
```
