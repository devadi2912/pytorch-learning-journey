**Comparison operators on tensors**

```python
import torch

print (torch.__version__)
if torch.cuda.is_available():
  print ("using gpu :", torch.cuda.get_device(0))
else :
  print ("gpu not available using cpu!")
```

```python
# tensor creation
f = torch.randint(size=(3,4), low=0 , high=100)
g = torch.randint(size=(3,4), low=0 , high=100)
print (f, g, sep="\n")
```

```python
# greater than
f > g
```

```python
# less than
f < g
```

```python
# equal to
f == g
```

```python
# not equal to
f != g
```

```python
# greater than equal to
f >= g
```

```python
# less than equal to
print (f <= g )
(f == g).dtype
```
