**Inplace Operations on tensors (saving compute and memory)**

```python
import torch
print (torch.__version__)

if torch.cuda.is_available():
  print ("gpu is available :")
  print ("using gpu :", torch.cuda.get_device_name(0))
else :
  print ("using cpu instead :")
```

```python
m = torch.rand(3,3)
n = torch.rand(3,3)

print (m,n, sep="\n")
```

```python
m = torch.transpose(m, 0, 1)
print (m)
m = torch.inverse(m)
print (m)
```

```python
print (torch.relu(m))

# this creates a completely new tensor of the same shame with the same number of dimentions
# to stop this form happening and to ensure that the we dont waste memory place we use inplace operations
```

implace operations ensure that the final result is stored in the same locaiton and not in any newer memory place. in pytorch inplace operations are easily denoted with a trailing "_" for example

```python
m + n # noraml additions will create a new tensor of the same shape somewhere in the memory
```

```python
# inplace addition
m.add_(n)
print (m)

# inplace relu
m.relu_()
print (m)
```

```python
# some other inplace operations
print (m.add_(n), end="\n\n")                       # add
print (m.sub(n), end="\n\n")                        # sub
print (m.mul_(n), end="\n\n")                       # multiply
print (m.div_(n), end="\n\n")                       # divide
print (m.round_(), end="\n\n")                      # round
print (m.floor_(), end="\n\n")                      # floor
print (m.ceil_(), end="\n\n")                       # ceil
print ((m*100).clamp_(min=20, max=30), end="\n\n")  # clamp
```
