**Mathematical operations one tensors**

```python
import torch

print (torch.__version__)
if torch.cuda.is_available():
  print ("using gpu :", torch.cuda.get_device(0))
else :
  print ("gpu not available using cpu!")
```

1. Scalar Operations on Tensors

```python
# scalar operations on the tensor x
x = torch.rand(2,4)
print (x, end="\n\n")

# addition
print (x + 7, end="\n\n")

# substraction
print (x - 2, end="\n\n")

# multiplication
print (x * 100, end="\n\n")

# division
print (x / 3, end="\n\n")

# int division
print ((x * 100)//3, end="\n\n")

# mod
print (((x * 100)//3)%2, end="\n\n")

# power
print (x ** 2, end="\n\n")
```

```python
# one example to get random value between 100 use
(x * 100).to(torch.int32).to(torch.float32)
```

2. Element Wise Operaitons

```python
a = torch.rand(3,2)
b = torch.rand_like(a)

# element wise operations on tensors

# addition
print (a + b, end="\n\n")
# substractions
print (a - b , end="\n\n")
# multiplication
print (a * b, end="\n\n")
# division
print (a / b, end="\n\n")
# int division
print ((a * b * 100)//2, end="\n\n")
# mod
print (a%b, end="\n\n")
# power
print (a**b, end="\n\n")
```

3. Reduction operations

```python
e = torch.randint(size=d.shape, low=0, high=100)
print (e)
e = torch.randint(size=(3,3), low=0, high=100)
print (e)
# reductions of tensors to a single value for simplifications will often be used in neural network building
```

```python
print (e, end="\n\n")

# sum
print (torch.sum (e), end="\n\n")   # reduce the entire tensor to a single sclar value that is the sum of all its values

# sum on column
print (torch.sum(e, dim=0), end="\n\n")  # dimention 0 indicates column wise operations

# sum on row
print (torch.sum(e, dim=1), end="\n\n")  # dimention 1 indicates row wise operations
```

```python
from sys import exception
# mean
try :
  print (torch.mean(e))
except Exception as err :
  print (err)
finally :
  # mean works only one tensors with float values
  f = torch.rand(3,3) * 20
  print (torch.mean(f))

print (f)
```

```python
# mean
print (torch.mean(f))

# mean on columns and on rows
print (torch.mean(f, dim=0) , torch.mean(f, dim=1)) # dim = 0 for columns and dim = 1 for rows
```

```python
# median
print (torch.median(f))

# median on columns and on rows
print (torch.median(f, dim=0)) # dim = 0 for columns and dim = 1 for rows
print (torch.median(f, dim=1))
```

```python
# max and min
print (f)

print (torch.max(f))  # returns the max value in the tensor
print (torch.min(f))  # returns the min value in the tensor
print (torch.max(f, dim=0).values)  # column wise max value in the tensor
print (torch.max(f, dim=1).values)  # column wise min value in their tensor
print (torch.min(f, dim=0).values)  # row wise max values in the tensor
print (torch.min(f, dim=1).values, end ="\n\n")  # row wise minimun values of the tensor

# indices
print (torch.max(f, dim=0).indices)  # column wise max value indices in the tensor
print (torch.max(f, dim=1).indices)  # column wise min value
print (torch.min(f, dim=0).indices)  # row wise max values indices in the tensor
print (torch.min(f, dim=1).indices)  # row wise minimun values
```

```python
# product
print (torch.prod(e), torch.prod(e, dim=0), torch.prod(e, dim=1), sep="\n")
```

```python
# standad deviation
# supports only floating point values
torch.std(f) # again can work for both rows and columns
```

```python
# variance
torch.var(f)
```

```python
# argmax and argmin
# return the position of the greatest and the smallest value in the tensor
print (torch.argmin(f))
print (torch.argmax(f))

# row and column wise positions
print (torch.argmin(f, dim=0))
print (torch.argmin(f, dim=1))
print (torch.argmax(f, dim=0))
print (torch.argmax(f, dim=1))
```

4. Matrix operations

```python
# matrices
f = torch.randint(size=(2,3), low=0, high=10)
g = torch.randint(size=(3,2), low=0, high=10)

print (f, g, sep="\n")
```

```python
# matrix multiplication
print (temp := torch.matmul(f, g))  # using warlus operator in python to assign values in expression
print (temp.shape)
```

```python
# dot product
vector1 = torch.tensor([1,2,3])
vector2 = torch.tensor([4,7,5])

print (torch.dot(vector1, vector2))
```

```python
h =  torch.rand(5,5)*10
# determinant and inverse only work for square matrices
```

```python
# determinant and inverse
print (torch.det(h)) # single value
print ()
print (torch.inverse(h)) # finds the inverse of the matrix h
```

5. Comparison operators on tensors

```python
# tensor creation
f = torch.randint(size=(3,4), low=0 , high=100)
g = torch.randint(size=(3,4), low=0 , high=100)
print (f, g, sep="\n")
```

```python
# greater than
f > g
```

```python
# less than
f < g
```

```python
# equal to
f == g
```

```python
# not equal to
f != g
```

```python
# greater than equal to
f >= g
```

```python
# less than equal to
print (f <= g )
(f == g).dtype
```
