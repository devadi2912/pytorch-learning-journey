```python
# requires_grad
"""
this is an attribute of a tensor in pytorch that when set to True marks a tesnor for derivative calcuations and pytorch internally maintains
a computation graph for that tesnor with all the opreations that are being perfromed on it in every step henceforth
"""

x = torch.tensor([3.0], requires_grad=True)  # this is a scalar quantity a tensor with 0 dimentions
print (x)
```

```python
y = x**2 # PowBackwards in a log of the power function that indicates that x is reainse to the power of two
# this record is stored in the computation graph
print (y)
```

**final derivation calculation**

```python
# calculating dy/dx
y.backward() # implementing chain rule calculus
```

```python
x.grad

# grad is an attribute that shows the final result after the derivative has been calcualted backwards througth teh computation graph
```
