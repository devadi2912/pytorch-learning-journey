```python
import torch
print (torch.__version__)

if torch.cuda.is_available() :
  print ("gpu is available:", torch.cuda.get_device_name(0))
else :
  print ("using cpu instead :")
```

```python
# binary cross entropy loss function .
def binary_cross_entropy_loss(prediction, target):
  epsilon = 1e-8
  prediction = torch.clamp(prediction, epsilon, 1-epsilon)  # to prevent log(0)
  loss = - (target * torch.log(prediction) + (1-target) * torch.log(1-prediction))
  return loss
```

```python
# inputs for the perceptron
x = torch.tensor(6.7)  # the input marks for the neruon
y = torch.tensor(0.0)  # the target marks to be achieved

w = torch.tensor(1.0)  # weigth of the perceptron
b = torch.tensor(0.0)  # bias of the perceptron

print (x,y,w,b,sep="\n")
```

**forward pass**

```python
# calculating the valus of x first accounting for the weigths and the biases
z = w * x + b

# calculating y prediction and then the loss
y_pred = torch.sigmoid(z)

# loss functions that is defined earlier in the code block in cell 2
loss = binary_cross_entropy_loss(y_pred,y)
print (y_pred,loss,sep="\n")
print ()
print (f"predictions : {y_pred}\nloss : {loss}")
```

```python
# derivative calculation
"""hardcoding the formula for each derivative"""

dloss_dy_pred = (y_pred - y) / (y_pred * (1 - y_pred))
dy_pred_dz = y_pred * (1 - y_pred)

dz_dw = x
dz_db = 1

# final loss gradient with respect to w and b
dloss_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dloss_db = dloss_dy_pred * dy_pred_dz * dz_db

print (f"{dloss_dw} : gradient of loss with respect to w")
print (f"{dloss_db} : gradient of loss with respect to b")
```
