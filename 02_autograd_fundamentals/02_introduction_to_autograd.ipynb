**Simplicity in autograd for derivative and gradient calcualtion**

```python
import torch

print (torch.__version__)

if torch.cuda.is_available():
  print ("gpu is available :", torch.cuda.get_device_name(0))
else :
  print ("using cpu instead :")
```

**one very important note is that the backward() function works only for a scalr and not for a tensor**

```python
"""
x = a ^ 2
y = sigmoid(x)
z = sin(y)
"""

def dx_da (a):
  return 2 * a

def dy_dx (x):
  return x.sigmoid()

def dz_dy (y):
  return y.cos()

torch.manual_seed(300)
a = torch.rand((1))
x = a**2
y = x.sigmoid()
z = y.sin()

print (x)
print (y)
print (z)

# finding derivatives of dz with respect to a is
"""dz/dx = dz/dy * dy/dx * dx/da"""
dz_da = dz_dy(y) * dy_dx(x) * dx_da(a)
print ()
print(dz_da )
```

```python
# with autograd
"""
x = a ^ 2
y = sigmoid(x)
z = sin(y)
"""

torch.manual_seed(300)
a = torch.rand((1), requires_grad=True )
x = a**2
y = x.sigmoid()
z = y.sin()

print (x)
print (y)
print (z)

# finding derivatives of dz with respect to a is
"""dz/dx = dz/dy * dy/dx * dx/da"""
z.backward()

print ()
print(a.grad)
```
