**perceptron calculations manualy (manual derivative calculation)**

```python
import torch
print (torch.__version__)

if torch.cuda.is_available() :
  print ("gpu is available:", torch.cuda.get_device_name(0))
else :
  print ("using cpu instead :")
```

```python
# binary cross entropy loss function .
def binary_cross_entropy_loss(prediction, target):
  epsilon = 1e-8 # to prevent log(0)
  prediction = torch.clamp(prediction, epsilon, 1-epsilon)
  loss = - (target * torch.log(prediction) + (1-target)*torch.lof(1-prediction))
  return loss
```

```python
# using the standard formual for the preceptron calculations
"""
z = w.x + b
y = sigmoid(z)
loss = -[y_target * ln(y_pred) + (1 - y_target) * ln(1 - y_pred)]
"""
```

```python
# inputs for the perceptron
x = torch.tensor(6.7)  # the input marks for the neruon
y = torch.tensor(0.0)  # the target marks to be achieved

w = torch.tensor(1.0, requires_grad=True )  # weigth of the perceptron || final loss gradient is found with respect to weigths and biases
b = torch.tensor(0.0, requires_grad=True )  # bias of the perceptron   || final loss gradient is found with respect to weigths and biases

print (x,y,w,b,sep="\n")
```

**forward pass**

```python
# calcualtion of z forward propagation
z = (w * x) + b

# final preictions usign sigmoid
y_pred = torch.sigmoid(z)

# calculation of the loss funcntion
loss = binary_cross_entropy_loss(y_pred, y)

print (y,loss,sep="\n")
print ()
print (f"{y_pred} : predictions \n{loss} : loss ")
```

```python
# back propagation
loss.backward()

# finding the gradients of loss with respect to loss

print (f"{w.grad} : gradient of loss with respect to w")
print (f"{b.grad} : gradient of loss with respect to b")
```
